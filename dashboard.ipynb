```python
df = pd.read_csv("CII_Master_Match_Level_Dataset (2).csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("Columns in the dataset:")
print(df.columns.tolist())

print("\nSelected captains:")
print(df["Captain"].unique())
```

```python
selected_captains = [
    "Sourav Ganguly",
    "Rahul Dravid",
    "MS Dhoni",
    "Virat Kohli",
    "Rohit Sharma"
]

dashboard_df = df[df["Captain"].isin(selected_captains)].copy()

print("Dashboard dataset created successfully!")
print("Total matches:", len(dashboard_df))
print("\nCaptains:")
print(dashboard_df["Captain"].value_counts())
```

```python
cii_summary = (
    dashboard_df[
        [
            "Captain",
            "Win_Score",
            "Pressure_Score",
            "Overseas_Score",
            "Toss_Decision_Score",
            "Home_Performance_Score",
            "CII"
        ]
    ]
    .drop_duplicates("Captain")
    .sort_values("CII", ascending=False)
    .reset_index(drop=True)
)

cii_summary
```

```python
fig = px.bar(
    cii_summary,
    x="Captain",
    y="CII",
    title="Captain Impact Index (CII) Comparison",
    text="CII",
    labels={
        "Captain": "Captain",
        "CII": "CII Score"
    }
)

fig.update_traces(
    texttemplate="%{text:.2f}",
    textposition="outside"
)

fig.update_layout(
    yaxis_range=[0, 100],
    height=500
)

fig.show()
```

```python
criteria_df = cii_summary[
    [
        "Captain",
        "Win_Score",
        "Pressure_Score",
        "Overseas_Score",
        "Toss_Decision_Score",
        "Home_Performance_Score"
    ]
].copy()

criteria_df
```

```python
criteria_long = criteria_df.melt(
    id_vars="Captain",
    var_name="Criteria",
    value_name="Score"
)

criteria_long["Criteria"] = criteria_long["Criteria"].replace({
    "Win_Score": "Win Score",
    "Pressure_Score": "Pressure Score",
    "Overseas_Score": "Overseas Score",
    "Toss_Decision_Score": "Toss Decision Score",
    "Home_Performance_Score": "Home Performance Score"
})

fig = px.bar(
    criteria_long,
    x="Criteria",
    y="Score",
    color="Captain",
    barmode="group",
    title="CII Criteria Comparison",
    text="Score"
)

fig.update_traces(
    texttemplate="%{text:.1f}",
    textposition="outside"
)

fig.update_layout(
    yaxis_range=[0, 100],
    height=600,
    xaxis_title="Performance Criteria",
    yaxis_title="Score"
)

fig.show()
```

```python
%%writefile cii_dashboard.py

import streamlit as st
import pandas as pd
import plotly.express as px

# Page configuration
st.set_page_config(
    page_title="Captain Impact Index",
    page_icon="🏏",
    layout="wide"
)

# Load dataset
df = pd.read_csv("CII_Master_Match_Level_Dataset.csv")

# Selected captains
selected_captains = [
    "Sourav Ganguly",
    "Rahul Dravid",
    "MS Dhoni",
    "Virat Kohli",
    "Rohit Sharma"
]

dashboard_df = df[df["Captain"].isin(selected_captains)].copy()

# Title
st.title("🏏 Captain Impact Index (CII)")
st.subheader("Indian Men's ODI Captains — 2005 to 2025")

st.divider()

# KPI calculations
total_matches = len(dashboard_df)
total_captains = len(selected_captains)
total_criteria = 5
analysis_period = "2005–2025"

# KPI cards
col1, col2, col3, col4 = st.columns(4)

with col1:
    st.metric("Total ODI Matches", total_matches)

with col2:
    st.metric("Captains Analyzed", total_captains)

with col3:
    st.metric("CII Criteria", total_criteria)

with col4:
    st.metric("Analysis Period", analysis_period)
```

```python
with open("cii_dashboard.py", "a") as f:
    f.write("""

# Home vs Overseas Analysis
st.divider()
st.header("🌍 Home vs Overseas Performance")

location_summary = (
    dashboard_df
    .groupby(["Captain", "Location_Type"])["India_Result"]
    .apply(lambda x: (x == "Win").sum() / ((x == "Win").sum() + (x == "Loss").sum()) * 100
           if ((x == "Win").sum() + (x == "Loss").sum()) > 0 else 0)
    .reset_index(name="Win_Percentage")
)

fig3 = px.bar(
    location_summary,
    x="Captain",
    y="Win_Percentage",
    color="Location_Type",
    barmode="group",
    text="Win_Percentage",
    title="Win Percentage: Home vs Overseas"
)

fig3.update_traces(
    texttemplate="%{text:.1f}%",
    textposition="outside"
)

fig3.update_layout(
    yaxis_range=[0, 100],
    height=550,
    xaxis_title="Captain",
    yaxis_title="Win Percentage (%)"
)

st.plotly_chart(fig3, use_container_width=True)
""")

print("Home vs Overseas Plotly chart added successfully!")
```

```python
with open("cii_dashboard.py", "a") as f:
    f.write("""

# CII Methodology
st.divider()
st.header("📘 CII Methodology")

st.write(
    "Captain Impact Index (CII) is a custom performance index developed "
    "for this project to analyze Indian Men's ODI captains from 2005 to 2025."
)

st.markdown("### CII Criteria and Weights")

methodology_data = pd.DataFrame({
    "Criterion": [
        "Win Score",
        "Pressure Score",
        "Overseas Score",
        "Toss Decision Score",
        "Home Performance Score"
    ],
    "Weight": [
        "25%",
        "25%",
        "20%",
        "15%",
        "15%"
    ]
})

st.dataframe(
    methodology_data,
    use_container_width=True,
    hide_index=True
)

st.markdown("### CII Formula")

st.code(
    "CII = 0.25 × Win Score + "
    "0.25 × Pressure Score + "
    "0.20 × Overseas Score + "
    "0.15 × Toss Decision Score + "
    "0.15 × Home Performance Score"
)

st.info(
    "Note: CII is a project-defined composite index and is not an official cricket statistic."
)
""")

print("CII methodology section added successfully!")
```
